### Problem 2 Quadratic equations

a) Escreva um programa que recebe como input tres numeros a, b, e c, e imprime as duas
solucoes da equacao quadratica $ax^{2} + bx + c = 0$ usando a formula usual

$$ x = \frac{-b \pm \sqrt{b^{2}-4ac}}{2a} $$

Use seu programa para calcular as solucoes de $0.001x^{2} +1000x +0.001 = 0$.

```python
import numpy as np
```

```python
a = float(input('a:'))
b = float(input('b:'))
c = float(input('c:'))

delta = np.pow(b, 2) - (4*a*c)

if delta < 0:
    print('delta is zero, try again')
else:
    x1 = - b/(2*a) + (delta**(1/2))/(2*a)
    x2 = - b/(2*a) - (delta**(1/2))/(2*a)

    print(f'x1 = {x1} and x2 = {x2}')
```

### Problem 3 Calculatting derivatives

Suponha que nós tenhamos uma função $f(x)$ e que desejamos calcular sua derivada em um ponto $x$. Nós podemos fazer isto com papel e caneta, se soubermos a fórmula matemática da função, ou podemos fazer isto com o computador, usando a definição de derivada:

$$ \frac{df}{dx} = \lim_{\delta \rightarrow 0} \frac{f(x+\delta) - f(x)}{\delta} $$

No computador, não podemos atingir o limite em que $\delta$ vai a zero, mas nós podemos conseguir uma aproximação razoável fazendo $\delta$ pequeno.

a) Escreva um programa que define uma função $f(x)$ que retorna o valor $x(x - 1)$, e então calcula a derivada da função no ponto $x = 1$ usando a fórmula acima com $\delta = 10^{-2}$.

Calcule o valor correto da mesma derivada analiticamente e compare com a resposta que seu programa forneceu. As duas não vão concordar perfeitamente. Por quê?

```python
def f(x):
    return x*(x-1)

def derivative(x, delta):
    
    dfdx = (f(x+delta) - f(x))/delta
    return dfdx
```

```python
delta = 1e-2
x = 1

result = derivative(x, delta)
result
```

b) Repita o cálculo para $\delta = 10^{-4}, 10^{-6}, 10^{-8}, 10^{-10}, 10^{-12}$ e $10^{-14}$. Você deve observar que a precisão do cálculo inicialmente melhora à medida que $\delta$ fica menor, mas que depois volta a piorar. Por que isto ocorre?

```python
deltas = [1e-4, 1e-6, 1e-8, 1e-10, 1e-12, 1e-14]

print('results and rests:\n')

for d in deltas:
    result = derivative(x, d)
    rest = 1-result
    
    print(f'{result} -> rest: {abs(rest):.2e}')
```
